```python
import pandas as pd
```

```python
# 1. Load original dataset
Sales = pd.read_csv("Sales.csv")
#Sales["TransactionDate"].unique()
# 2. Convert original dates (DD/MM/YYYY)
Sales["TransactionDate"] = pd.to_datetime(Sales["TransactionDate"], format ="%d/%m/%Y")
#convert as dateformat m/d/y
Sales["TransactionDate"] = Sales["TransactionDate"].dt.strftime("%m/%d/%Y")
Sales["TransactionDate"] = pd.to_datetime(Sales["TransactionDate"], format ="%m/%d/%Y")
# 3. Filter missing TransactionID &  New dataset that we use tu work is Sales_Filtered
Sales_Filtered = Sales.dropna(subset=["TransactionID"])
Sales_Filtered["TransactionID"] = Sales_Filtered["TransactionID"].astype(int)
#Sales_Filtered.head()
# Which values contain letters
#Sales_Filtered[
#    ~Sales_Filtered["SalesAmount"].str.fullmatch(r"\d+(\.\d+)?", na=False)
#]["SalesAmount"].unique()
Sales_Filtered["SalesAmount"] = Sales_Filtered["SalesAmount"].replace({
    "two-hundred": 200,
    "one-hundred": 100,
    "four-hundred": 400,
    "six-hundred": 600
})
#Sales_Filtered.head()
#Step 3. Convert the entire column to numeric
Sales_Filtered["SalesAmount"] = pd.to_numeric(
    Sales_Filtered["SalesAmount"],
    errors="raise"
)
print(Sales_Filtered.isna().sum())
print(Sales_Filtered.dtypes)
print(Sales_Filtered["SalesAmount"].sum())
```

```python
def convert_date(value):
    first, second, year = map(int, value.split("/"))

    if first > 12:
        return pd.to_datetime(value, format="%d/%m/%Y")

    elif second > 12:
        return pd.to_datetime(value, format="%m/%d/%Y")

    else:
        return pd.NaT
# MAP FUNCTION IS TO APPLY METHOD ALL THE LIST DICTIONARIO TUPPLES.
```

```python
#WORK WITH INVENTORY FILE
import pandas as pd
inventory = pd.read_csv("Inventory.csv")
#inventory.info()
#inventory.head()
#First we gonna work indentifying date patterns creating a new column
inventory["New_Column"] = inventory["LastRestocked"].apply(convert_date)
#inventory.head()
#inventory["New_Column"].isna().sum()
inventory["New_Column"] = pd.to_datetime(
    inventory["LastRestocked"],
    format="%d/%m/%Y",
    errors="coerce"
)
inventory["LastRestocked"] = inventory["New_Column"]
inventory = inventory.drop(columns=["New_Column"]) ## DELETE THE COLUMN WE CREATE TO MAKE VALIDATION
inventory["Category"] = inventory["Category"].fillna(
    "No Category Provided"
)
inventory["Category"] = inventory["Category"].astype("category")
#inventory.info()
print(inventory.isna().sum())
inventory[inventory["ProductName"].isna()]
inventory_filtered = inventory.dropna(subset = ["ProductName"])
#inventory_filtered
inventory_filtered["RestockingFrequency"] = inventory_filtered["RestockingFrequency"].replace({"30 d" : "30 days"}) 
inventory_filtered.info()
```
